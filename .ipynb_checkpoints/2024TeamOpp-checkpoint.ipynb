```python
from bs4 import BeautifulSoup, Comment
import pandas as pd

import requests
```

```python
#urls

url1 = 'https://www.basketball-reference.com/teams/BOS/2024.html'
url2 = 'https://www.basketball-reference.com/teams/GSW/2024.html'
page = requests.get(url1)

soup = BeautifulSoup(page.text, 'html')

# print(soup)
```

```python
#Creating Main DF
df = pd.DataFrame(columns = ['Name', 'MP', 'FG', 'FGA', 'FG%', '3P', '3PA', '3P%', '2P', '2PA', '2P%', 'FT', 'FTA', 'FT%', 'ORB', 'DRB', 'TRB', 'AST', 'STL', 'BLK', 'TOV', 'PF', 'PTS'])
print(df)
```

```python
def getData(df, url):

    """
    Given a url and the df, grab team stats and append to df
    """

    # ===== TODO: get soup variable of url ===== #
    page = requests.get(url)
    soup = BeautifulSoup(page.text, 'html')
    # ===== End of TODO: get soup variable of url ===== #

   

    # ===== TODO: get team name and year ===== #
    info_tag = soup.find('div', id='info')
    span_tags = info_tag.find_all('span')
    name = span_tags[0].text + ' ' + span_tags[1].text
    # ===== End of TODO: get team name and year ===== #


    # ===== TODO: parse section that's commented out in BBRef ===== #
    div_tag = soup.find('div', id='all_team_and_opponent')    
    comments = div_tag.find_all(string=lambda text: isinstance(text, Comment))
    
    for comment in comments:
        if 'div_team_and_opponent' in comment:
            commented_soup = BeautifulSoup(comment, 'html.parser')
            div_tag = commented_soup.find('div', id='div_team_and_opponent')

    tbody = div_tag.find_all('tbody')
    tr = tbody[0].find_all('tr')
    # ===== End of TODO: parse section that's commented out in BBRef ===== #

    
    
    # ===== TODO: parse team stats and append to df ===== #
    for i in range(len(tr)):
        td = tr[i].find_all('td')
        row = [cell.text.strip() for cell in td]    
        for j in range(len(row)):
            df.loc[i, df.columns[j + 1]] = row[j]
    
    list = df.iloc[1].tolist()
    list[0] = name
    del list[1]
    # ===== End of TODO: parse team stats and append to df ===== #
        
```
