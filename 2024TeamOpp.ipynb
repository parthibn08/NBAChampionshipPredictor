```python
from bs4 import BeautifulSoup, Comment
import pandas as pd

import requests
```

```python
#urls

url1 = 'https://www.basketball-reference.com/teams/BOS/2024.html'
url2 = 'https://www.basketball-reference.com/teams/BOS/2023.html'
url3 = 'https://www.basketball-reference.com/teams/BOS/2022.html'
url4 = 'https://www.basketball-reference.com/teams/BOS/2021.html'
url5 = 'https://www.basketball-reference.com/teams/BOS/2020.html'
url6 = 'https://www.basketball-reference.com/teams/BOS/2019.html'
url7 = 'https://www.basketball-reference.com/teams/BOS/2018.html'
url8 = 'https://www.basketball-reference.com/teams/BOS/2017.html'
url9 = 'https://www.basketball-reference.com/teams/BOS/2016.html'
url10 = 'https://www.basketball-reference.com/teams/BOS/2015.html'
url11 = 'https://www.basketball-reference.com/teams/BOS/2014.html'
url12 = 'https://www.basketball-reference.com/teams/BOS/2013.html'
url13 = 'https://www.basketball-reference.com/teams/BOS/2012.html'

urList = [url1, url2, url3, url4, url5, url6, url7, url8, url9, url10, url11, url12, url13]
```

```python
#Creating Main DF
MainDF = pd.DataFrame(columns = ['Name', 'MP', 'FG', 'FGA', 'FG%', '3P', '3PA', '3P%', '2P', '2PA', '2P%', 'FT', 'FTA', 'FT%', 'ORB', 'DRB', 'TRB', 'AST', 'STL', 'BLK', 'TOV', 'PF', 'PTS'])
print(MainDF)
```

```python
#Concatenating all team data into one DF
for i in range(len(urList)):
    list = getData(urList[i])
    MainDF.loc[len(MainDF)] = list

print(MainDF)
```

```python
def getData(url):

    """
    Given a url and the df, grab team stats and append to df
    """

    # ===== TODO: get soup variable of url ===== #
    page = requests.get(url)
    soup = BeautifulSoup(page.text, 'html')
    # ===== End of TODO: get soup variable of url ===== #

    

    # ===== TODO: get team name and year ===== #
    info_tag = soup.find('div', id='info')
    span_tags = info_tag.find_all('span')
    name = span_tags[0].text + ' ' + span_tags[1].text
    # ===== End of TODO: get team name and year ===== #


    # ===== TODO: parse section that's commented out in BBRef ===== #
    div_tag = soup.find('div', id='all_team_and_opponent')    
    comments = div_tag.find_all(string=lambda text: isinstance(text, Comment))
    
    for comment in comments:
        if 'div_team_and_opponent' in comment:
            commented_soup = BeautifulSoup(comment, 'html.parser')
            div_tag = commented_soup.find('div', id='div_team_and_opponent')

    tbody = div_tag.find_all('tbody')
    tr = tbody[0].find_all('tr')
    # ===== End of TODO: parse section that's commented out in BBRef ===== #

    # ===== TODO: getting stat header row and turning into df ===== # 
    stats = div_tag.find_all('thead')
    th = stats[0].find_all('th')
    stats_row = [stat.text for stat in th]
    stats_row[0] = 'Blank' 
    df = pd.DataFrame(columns = stats_row)
    # ===== End of TODO: getting stat header row and turning into df ===== #

    # ===== TODO: add team entry ===== #
    first_column_entries = ['Team']
    for i, value in enumerate(first_column_entries):
        df.loc[i, 'Blank'] = value
    # ===== End of TODO: add team entry ===== #
   
    
    # ===== TODO: parse team stats===== #
    for i in range(len(tr)):
        td = tr[i].find_all('td')
        row = [cell.text.strip() for cell in td]    
        for j in range(len(row)):
            df.loc[i, df.columns[j + 1]] = row[j]
    
    list = df.iloc[1].tolist()
    list[0] = name
    del list[1]
    return list
    # ===== End of TODO: parse team stats===== #
        
```
